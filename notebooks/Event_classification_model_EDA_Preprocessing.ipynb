#For the cols:1-team_centroid_distance 2-dist_nearest_defender 3_open_goal_angle these will be dropped because they are only known for the shot rows because they need the opponent locations which needs the shot freeze fram which is only know for shots

```python
df=df.drop(['team_centroid_distance','dist_nearest_defender','open_angle_goal'],axis=1)
```

#Ball speed cannot be drooped because only 37% missing so it needs to be imputed but as we see that ball speed feature is MAR which means its missingnes is due to other event classes types which are foul and other classes so we will group by event_class type and impute with median and at the same time we will add a ball_missingness column to tell the model that even the imputed value in other classes was actually missing values

```python
df['ball_speed_missing'] = df['ball_speed'].isnull().astype(int)

df['ball_speed'] = df.groupby('event_class')['ball_speed'].transform(lambda x: x.fillna(x.median()))
```

#Trajectory also will not be dropped as only 17% missing which are also MAR due to other,Foul classes and it is logical as those classes don't have any relation with the ball to have trajectories so will just fill the nulls with 0 and make a trajectory_missing column indication that these zeros represent missing trajectories

```python
df['trajectory_missing'] = df['trajectory_length'].isnull().astype(int)
df['trajectory_length'] = df['trajectory_length'].fillna(0)
df['trajectory_angle'] = df['trajectory_angle'].fillna(0)
```

#the remaining 47 missing geomtrical featues are missing in the other class and only represents less than 0.01$ of total rows so they will be dropped

```python
df =df.dropna(subset=['ball_x_start', 'ball_y_start', 'dist_to_goal', 'shot_angle']).reset_index(drop=True)
```
